# Registrierungen

Tagesliste aus `USER_REGISTRATION_DB.ANALYTICS.DAILY_REGISTRATIONS_TOTAL`.

Zum Aktualisieren **Run all** ausfuehren. Fuer eine Ansicht, die sich selbst
aktualisiert, das Streamlit-Dashboard `REGISTRATION_DASHBOARD` verwenden.

```python
from snowflake.snowpark.context import get_active_session

session = get_active_session()

daily = session.sql(
    """
    SELECT REGISTRATION_DATE, NEW_USERS
    FROM USER_REGISTRATION_DB.ANALYTICS.DAILY_REGISTRATIONS_TOTAL
    ORDER BY REGISTRATION_DATE DESC
    """
).to_pandas()

daily
```

```python
recent = session.sql(
    """
    SELECT USER_ID, REGISTERED_AT
    FROM USER_REGISTRATION_DB.ANALYTICS.REGISTRATIONS
    ORDER BY REGISTERED_AT DESC
    LIMIT 25
    """
).to_pandas()

recent
```
